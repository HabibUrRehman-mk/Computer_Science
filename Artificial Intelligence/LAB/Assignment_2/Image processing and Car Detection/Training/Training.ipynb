```python
from ultralytics import YOLO

model = YOLO("yolov8s.pt")

# Start training
results = model.train(
    data="Cars Detection/data.yaml",   
    epochs=50,                         
    imgsz=640,
    batch=8,                           
    device=0,
    workers=2,                         
    cache=True,                        
    cos_lr=True,
    optimizer="AdamW",
    lr0=0.001,
    weight_decay=0.0005,
    augment=True,
    auto_augment='randaugment',
    amp=True,                          
    patience=30,
    project="runs/train",
    name="car_detector_colab",
    verbose=True,
)
```

```python
# Load best weights
model = YOLO("/content/runs/train/car_detector_colab/weights/best.pt")
results = model.predict(source="/content/Cars Detection/test/images", conf=0.4, save=True)
```
